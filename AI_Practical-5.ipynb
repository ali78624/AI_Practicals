```python
def laptop_expert(turns_on, internet_works, is_slow):
    if turns_on == 'no':
        return "Check power supply and battery."
        
    elif internet_works == 'no':
        return "Check Wi-Fi connection and router settings."
        
    elif is_slow == 'yes':
        return "Consider clearing temporary files, checking for viruses, or upgrading RAM."
        
    else:
        return "Your laptop seems to be working fine. Enjoy!"

turns_on = input("Does the laptop turn on? yes/no: ").lower()
internet_works = input("Does the internet work? yes/no: ").lower()
is_slow = input("Is the laptop slow? yes/no: ").lower()
advice = laptop_expert(turns_on, internet_works, is_slow)
print("\nExpert System Advice:")
print(advice)
```
